```python
import numpy as np
from cuml.kernel_ridge import KernelRidge
import tqdm
```

```python
def loss_fn(x):
    return np.squeeze((x - 2)**2, axis=-1)
```

```python
model1 = KernelRidge(alpha=10.0, kernel="rbf")
model2 = KernelRidge(alpha=1.0, kernel="rbf")
rng = np.random.default_rng(12345)
```

```python
weighted = True
epsilon = 0.001
n_iters = 100
batch_size = 128
direction_size = 32
parameter_size = 1
```

```python
direction = rng.standard_normal((batch_size, direction_size))
output = rng.standard_normal((batch_size, parameter_size))
model1.fit(direction, output)
model2.fit(direction, output)
```

```python
for i in tqdm.trange(n_iters):
	direction = rng.standard_normal((batch_size, direction_size))
	candidate1 = model1.predict(direction)
	candidate2 = model2.predict(direction)
	loss1 = loss_fn(candidate1)
	loss2 = loss_fn(candidate2)
	best_candidate = np.where((loss1 < loss2).reshape(-1, 1), candidate1, candidate2)
	if weighted:
		best_loss = np.minimum(loss1, loss2)
		reduction1 = loss1 - best_loss
		reduction2 = loss2 - best_loss
		movement1 = np.linalg.norm(candidate1 - best_candidate, axis=1)
		movement2 = np.linalg.norm(candidate2 - best_candidate, axis=1)
		sample_weight1 = (reduction1 + epsilon) / (movement1 + epsilon)
		sample_weight2 = (reduction2 + epsilon) / (movement2 + epsilon)
		model1.fit(direction, best_candidate, sample_weight1)
		model2.fit(direction, best_candidate, sample_weight2)
	else:
		model1.fit(direction, best_candidate)
		model2.fit(direction, best_candidate)
```
